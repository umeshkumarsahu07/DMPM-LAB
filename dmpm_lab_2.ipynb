```python
n = list(map(int, input("Enter numbers: ").split()))

freq = {}

for i in n:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

max_freq = 0
mode = 0

for i in freq:
    if freq[i] > max_freq:
        max_freq = freq[i]
        mode = i

print("Mode =", mode)
```
