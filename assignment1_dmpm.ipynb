```python
n=int(input("enter the number"))
count=0
for i in range(1,n+1):
  if n%i==0:
    count+=1
if count<=2:
  print(f"the number {n} is prime")
else:
  print("not prime")
```

```python
def fact(num):
  factorial=1
  if num==1 or num==0:
    return factorial
  else:
    for i in range(1,num+1):
      factorial*=i
    return factorial

number=int(input("enter the number"))
print(fact(number))
```

```python
def fact(n):
  factorial=1
  if n==0 or n==1:
    return factorial
  else:
    return fact(n)*fact(n-1)
```
