```python

import matplotlib.pyplot as plt

subjects = ["Math", "Java", "Python", "DBMS"]
marks = [80, 75, 90, 85]

plt.bar(subjects, marks)
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Marks in Subjects")
plt.show()
```

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]

plt.plot(days, temperature, marker="o")
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Temperature for 5 Days")
plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ["Math", "Java", "Python", "DBMS"]
marks = [80, 75, 90, 85]

plt.pie(marks, labels=subjects, autopct="%1.1f%%")
plt.title("Marks Distribution")
plt.show()
```

```python
import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 60, 65, 70, 70, 75, 80, 85, 90]

plt.hist(marks, bins=5)
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Marks Distribution")
plt.show()
```

```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6]
marks = [40, 45, 55, 60, 70, 80]

plt.scatter(hours, marks)
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("StudentsPerformance.csv")

print(df.head())
```

```python
avg_marks = df.groupby("gender")["math score"].mean()

plt.bar(avg_marks.index, avg_marks.values)
plt.xlabel("Gender")
plt.ylabel("Average Math Score")
plt.title("Average Math Score by Gender")
plt.show()
```

```python
plt.plot(df["math score"].head(10), marker="o")
plt.xlabel("Student Number")
plt.ylabel("Math Score")
plt.title("Math Scores of First 10 Students")
plt.show()
```

```python
gender_count = df["gender"].value_counts()

plt.pie(gender_count.values, labels=gender_count.index, autopct="%1.1f%%")
plt.title("Gender Distribution")
plt.show()
```

```python
plt.hist(df["math score"], bins=10)
plt.xlabel("Math Score")
plt.ylabel("Number of Students")
plt.title("Distribution of Math Scores")
plt.show()
```

```python
plt.scatter(df["math score"], df["reading score"])
plt.xlabel("Math Score")
plt.ylabel("Reading Score")
plt.title("Math Score vs Reading Score")
plt.show()
```
