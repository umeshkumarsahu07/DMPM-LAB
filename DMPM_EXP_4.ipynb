```python
def min_max_normalize(data):
    min_val = min(data)
    max_val = max(data)

    if max_val == min_val:
        return [0 for _ in data]

    return [(x - min_val) / (max_val - min_val) for x in data]


# Example
data = [10, 20, 30, 40, 50]

normalized = min_max_normalize(data)

print(normalized)
```

```python
import pandas as pd

# Load dataset
df = pd.read_csv("diabaties.data.csv", header=None)

# Add column names
df.columns = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age",
    "Outcome"
]

print(df.head())
```

```python
# Features to normalize
features = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age"
]

# Min-Max normalization
df[features] = (
    df[features] - df[features].min()
) / (
    df[features].max() - df[features].min()
)

print(df.head())
```
