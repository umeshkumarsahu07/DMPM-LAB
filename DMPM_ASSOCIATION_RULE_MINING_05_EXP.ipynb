```python
# ============================================================
#          ASSOCIATION RULE MINING - APRIORI
# ============================================================

# Transactions
transactions = {
    "T1": {"bread", "milk"},
    "T2": {"bread", "diaper", "beer", "eggs"},
    "T3": {"milk", "diaper", "beer", "coke"},
    "T4": {"bread", "milk", "diaper", "beer"},
    "T5": {"bread", "milk", "diaper", "coke"}
}

# Minimum thresholds
MIN_SUPPORT = 60       # 60%
MIN_CONFIDENCE = 60    # 60%

total_transactions = len(transactions)

# Minimum support count
min_support_count = (MIN_SUPPORT / 100) * total_transactions


# ============================================================
# 1. FUNCTION TO FIND SUPPORT COUNT
# ============================================================

def support_count(itemset):
    count = 0

    for items in transactions.values():
        if itemset.issubset(items):
            count += 1

    return count


# ============================================================
# 2. FIND FREQUENT 1-ITEMSETS
# ============================================================

all_items = set()

for items in transactions.values():
    all_items.update(items)

frequent_items = {}

for item in sorted(all_items):
    count = support_count({item})

    if count >= min_support_count:
        frequent_items[item] = count


# ============================================================
# 3. DISPLAY FREQUENT 1-ITEMSETS
# ============================================================

print("=" * 60)
print("             FREQUENT 1-ITEMSETS")
print("=" * 60)

for item, count in frequent_items.items():
    support = (count / total_transactions) * 100

    print(f"{item:10} | Support Count = {count} | Support = {support:.0f}%")


# ============================================================
# 4. FIND FREQUENT 2-ITEMSETS
# ============================================================

items = list(frequent_items.keys())
frequent_pairs = {}

for i in range(len(items)):
    for j in range(i + 1, len(items)):

        pair = {items[i], items[j]}
        count = support_count(pair)

        if count >= min_support_count:
            frequent_pairs[frozenset(pair)] = count


# ============================================================
# 5. DISPLAY FREQUENT 2-ITEMSETS
# ============================================================

print("\n" + "=" * 60)
print("             FREQUENT 2-ITEMSETS")
print("=" * 60)

for pair, count in frequent_pairs.items():
    support = (count / total_transactions) * 100

    pair_name = " , ".join(sorted(pair))

    print(
        f"{pair_name:25} | "
        f"Support Count = {count} | "
        f"Support = {support:.0f}%"
    )


# ============================================================
# 6. GENERATE ASSOCIATION RULES
# ============================================================

print("\n" + "=" * 60)
print("             ASSOCIATION RULES")
print("=" * 60)

for pair, pair_count in frequent_pairs.items():

    pair = list(pair)

    A = {pair[0]}
    B = {pair[1]}

    # --------------------------------------------------------
    # Rule: A -> B
    # --------------------------------------------------------

    A_count = support_count(A)

    confidence = (pair_count / A_count) * 100

    if confidence >= MIN_CONFIDENCE:

        print(
            f"{pair[0]:10} -> {pair[1]:10} | "
            f"Support = {(pair_count / total_transactions) * 100:.0f}% | "
            f"Confidence = {confidence:.0f}%"
        )

    # --------------------------------------------------------
    # Rule: B -> A
    # --------------------------------------------------------

    B_count = support_count(B)

    confidence = (pair_count / B_count) * 100

    if confidence >= MIN_CONFIDENCE:

        print(
            f"{pair[1]:10} -> {pair[0]:10} | "
            f"Support = {(pair_count / total_transactions) * 100:.0f}% | "
            f"Confidence = {confidence:.0f}%"
        )


# ============================================================
# 7. SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("                    SUMMARY")
print("=" * 60)

print(f"Total Transactions       : {total_transactions}")
print(f"Minimum Support          : {MIN_SUPPORT}%")
print(f"Minimum Support Count    : {min_support_count:.0f}")
print(f"Minimum Confidence       : {MIN_CONFIDENCE}%")
print(f"Frequent 1-Itemsets      : {len(frequent_items)}")
print(f"Frequent 2-Itemsets      : {len(frequent_pairs)}")
print("=" * 60)
```

```python
# ============================================================
#       CHI-SQUARE TEST OF INDEPENDENCE
#       Gender vs Flavor Preference
# ============================================================

import numpy as np
from scipy.stats import chi2_contingency

# ------------------------------------------------------------
# 1. Observed Frequency Table
# ------------------------------------------------------------

observed = np.array([
    [23, 18,  8,  8],    # Male
    [15,  6, 12, 10]     # Female
])

rows = ["Male", "Female"]
columns = ["Strawberry", "Coffee", "Orange", "Vanilla"]

# ------------------------------------------------------------
# 2. Hypotheses
# ------------------------------------------------------------

print("=" * 65)
print("              CHI-SQUARE TEST OF INDEPENDENCE")
print("=" * 65)

print("\nH0 (Null Hypothesis):")
print("Gender and flavor preference are independent.")

print("\nH1 (Alternative Hypothesis):")
print("Gender and flavor preference are not independent.")


# ------------------------------------------------------------
# 3. Perform Chi-Square Test
# ------------------------------------------------------------

chi_square, p_value, degrees_of_freedom, expected = chi2_contingency(observed)


# ------------------------------------------------------------
# 4. Degrees of Freedom
# ------------------------------------------------------------

print("\n" + "-" * 65)
print("DEGREES OF FREEDOM")
print("-" * 65)

print("(Rows - 1) × (Columns - 1)")
print("(2 - 1) × (4 - 1)")
print("Degrees of Freedom =", degrees_of_freedom)


# ------------------------------------------------------------
# 5. Expected Frequency Table
# ------------------------------------------------------------

print("\n" + "-" * 65)
print("EXPECTED FREQUENCY TABLE")
print("-" * 65)

print(f"{'Gender':<12}", end="")

for column in columns:
    print(f"{column:<14}", end="")

print()

for i in range(len(rows)):
    print(f"{rows[i]:<12}", end="")

    for j in range(len(columns)):
        print(f"{expected[i][j]:<14.2f}", end="")

    print()


# ------------------------------------------------------------
# 6. Chi-Square Results
# ------------------------------------------------------------

print("\n" + "-" * 65)
print("CHI-SQUARE TEST RESULTS")
print("-" * 65)

print(f"Chi-Square Statistic : {chi_square:.4f}")
print(f"P-Value              : {p_value:.4f}")
print(f"Degrees of Freedom   : {degrees_of_freedom}")


# ------------------------------------------------------------
# 7. Conclusion
# ------------------------------------------------------------

alpha = 0.05

print("\n" + "-" * 65)
print("CONCLUSION")
print("-" * 65)

if p_value < alpha:
    print("Reject H0.")
    print("There is a significant association between")
    print("gender and flavor preference.")
else:
    print("Fail to reject H0.")
    print("There is no significant association between")
    print("gender and flavor preference.")

print("=" * 65)
```
